```python
#reading string values from user
s= input()
print(type(s),s)
print(int(s))
print(float(s))
print(list(s)) #It returns list of characters
print(tuple(s)) #It returns tuple of characters
print(set(s)) # It returns set of characters
```

```python
#reading tuple of integers
n = tuple(map(int,input().split()))
print(n)
print(list(n))
print(set(n))
print(str(n))
```

```python
#reading set of integers
n = set(map(int,input().split()))
print(n)
print(list(n))
print(tuple(n))
print(str(n))
```
