```python
import math
print(dir(math))
```

```python
print(math.ceil(4.95)) # it returns the upper bound value
print(math.floor(4.35)) # it returns the lower bound value
print(math.factorial(5)) # it returns factorial of a number
print(math.gcd(10,20,3)) # it returns the GCD value
print(math.lcm(10,40))
print(math.log10(2))
print(math.log10(4))
print(math.log2(4))
print(math.pi)
```

```python
import random
print(random.random()) # This function returns the value in between 0 to 1
print(random.randint(1,10)) # This function generates random number between a and b
print(random.choice(['A',1,2,3,4,5,6,7,'C','Q','D'])) # It returns the random item 
print(random.choices(['A',1,2,3,4,5,6,7,'C','Q','D'],k=3))
arr=['A',1,2,3,4,5,6,7,'C','Q','D']
random.shuffle(arr)
print(arr)
```
