```python
#check weather the number is even or odd without using "else" block
n=int(input())
if n % 2==0:
    print("even")
if n % 2==1:
    print("odd")

#check weather the number is divisible with 3 
n=int(input())
if n % 3==0:
    print("fizz")
if n % 5==0:
    print("buzz")
else:
    print(n)
```
