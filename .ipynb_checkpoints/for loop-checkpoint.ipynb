```python
#count vowels and consonents in string
string=input().lower()
c=v=0
for char in string:
    if char in 'aeiou':
        c=c+1
    else:
        v=v+1
print(c,v)
```

```python
#count vowels and consonents in string
string=input().lower()
c=v=0
for char in string:
    if char.isalpha():
        if char in 'aeiou':
            c=c+1
        else:
            v=v+1
print(c,v)
```

```python
#count even and odd numbers in given list of numbers
numbers = [1,2,3,4,5,6]
even_count = 0
odd_count = 0

for num in numbers:
    if num % 2 == 0:
        even_count += 1
    else:
        odd_count += 1

print("Even numbers count:", even_count)
print("Odd numbers count:", odd_count)
```

```python
#seperate the even and odd number in given list
numbers = [1,2,3,4,5,6]

even_numbers = [num for num in numbers if num % 2 == 0]
odd_numbers = [num for num in numbers if num % 2 != 0]

print("Even numbers:", even_numbers)
print("Odd numbers:", odd_numbers)
```
