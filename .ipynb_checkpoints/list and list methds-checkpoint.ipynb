```python
#list ordered collection
lst = [1,2,3,4.45,"hello"]
print(lst)
```

```python
# list is mutable
lst = [1,2,3,4.45,"hello"]
lst[2] = 100
print(lst)
```

```python
# indexing and slicing
lst = [1,2,3,4.45,"hello"]
print(lst[-2])
print(lst[-1])
print(lst[-3:])
print(lst[:100])
print(lst[:100:10])
print(lst[-10:-1:-1])
print(lst[-10:-1])
print(lst[::-1])
print(lst[::-3])
print(lst[-1:])
print(lst[-1:1:])
```

```python
# define empty list
l1 =[]
l2 =list()
print(type(l1),type(l2))
```

```python
# list Concatiantion
l1 = [1,2,3]
l2 = [2,3,4,5]
res = l1+l2
print(res)
print(l1+l1)

# list reperatation
print(11*3)
print([0] * 10)
print([]*10)

#add single element 10 to list
l1 = [1,2,3,4]
res =l1+[10]
print(res)
```

```python
## list methods
- Element adding Methods
 - append(item) : It adds item at end of the list
 -extend(itarable) : It adds all itarable elements at end of list
 -insert(post, item) : This Method adds the item at specified  the position
```

```python
lst =[1,2,3,4]
lst.append(10)
print(lst)
lst.append("code")
print(lst)
lst.append([10,20,30])
print(lst)
print(lst[5])
print(lst[1],[2])

lst.extend([10,20,30])
print(lst)

lst.insert(1,1000)
print(lst)
```

```python
# Element removing Methods
-remove(val): It removes the first occurence of value if value exists in list, otherwise it raises value error
-pop(): It remove and retrns the last of element of list
-pop(pos):It remove and returns the specified position element from list
-clear():It removes all elements from list
-del: It python keyword, used to delete values.
```

```python
lst = [10,20,30,40,20]
lst.remove(20)
print(lst)
#lst.remove(100)  # error
val=lst.pop()
print(val,lst)
val2=lst.pop(1)
print(val2,lst)

del lst[0]
print(lst)

lst.clear()
print(lst)
```

```python
# index(val): It returns the index of first occureence of value if value exists in list otherwise ValueError
# count(val): It retrns the total frequency of value
lst = [11,12,13,2,3,4,2,3,3,3]
ind = lst.index(3)
print(ind)
print(lst.count(3))
print(lst.count(100))
print(lst.index(100))
```

```python
#sort method(reverse = false): It sorts the all list elements in ascending order, if reverse value is true it sorts in decending order.

lst = [5,22,43,55,-1,-3,-4]
lst.sort()
print(lst)
lst.sort(reverse=True)
print(lst)
```

```python
# Built in functions 
# len(),min(),max(),sum()
lst = [1,2,3,4]
print(len(lst),min(lst),max(lst),sum(lst))

lst = ['All','How are you','Hello','hi','what','you','Here']
print(len(lst),min(lst),max(lst))
lst.sort()
print(lst)

```
