```python
#generator
def gen(x):
    x=x+1
    yield x
    x=x+2
    yield x
    x=x+3
    yield x
res = gen(5)
print(res)
print(next(res))
print(next(res))
print(next(res))
```

```python
# to overcome the stop iteration the error
def gen(x):
    x=x+1
    yield x
    x=x+2
    yield x
    x=x+3
    yield x
res = gen(5)
for i in res:
    print(i)
```

```python
# Generator function to yield numbers from 1 to 10
def gen():
    for i in range(1, 11):
        yield i
for j in gen():
    print(j)
```

```python
def gen_even(n):
    for i in range(2, n+1, 2):
        yield i
for num in gen_even(10):
    print(num)
```

```python
def gen_fib(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b
for num in gen_fib(7):
    print(num)
```
