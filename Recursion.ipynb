```python
def fun(n):
    if n == 1:
        return
    print(n)
    fun(n - 1)
fun(5)
```

```python
#find factorial of a number
def fact(n):
    if n == 0 or n == 1:
        return 1
    return n * fact(n - 1)
print(fact(6)) 
```

```python
#Fibonacci
def fib(n):
    if n == 0 or n == 1:
        return 1
    return  fib(n - 1)+ fib(n-2)
fib(6) 
```

```python
#sum of 'n' natural numbers using recursion
def sum_nat(n):
    if n == 0:
        return 0
    return n + sum_nat(n - 1)
sum_nat(5)
```

```python
#product of 'n' natural numbers using recursion
def pro_nat(n):
    if n == 0:
        return 0
    return n * pro_nat(n - 1)
pro_nat(1)
```

```python
#Find Binary number for decimal number by using recursion
def dec_to_bin(n):
    if n == 0:
        return "0"
    elif n == 1:
        return "1"
    return dec_to_bin(n // 2) + str(n % 2)
dec_to_bin(10)
```

```python
#Anonymous Functions: An anonymous function in Python is a function defined without a name,
            #created using the lambda keyword.
#syntax: lambda arguments: expression
add=lambda x: x+2
print(add(2))
```

```python
#Find a square of a number using lambda
square = lambda n: n * 4
print(square(4))
```

```python
#Sum of two numbers using lambda
add=lambda x,y:x+y
print(add(5,4))
```

```python
#Find square of all list of integers using lambda and map()
arr=[1,2,3,4,5]
res=list(map(lambda x:x**2, arr))
print(res)
```

```python
# Find the even number in the list using lambda and filter()
arr=[1,2,3,4,5]
res=list(filter(lambda x:x%2==0,arr))
print(res)
```

```python
#Single line if-else syntax
#True_exp if condition else False_exp
#Find the weather the number is even or odd using lambda
n=int(input())
res=lambda x:"odd" if x%2 else "even"
print(res(n))
```
