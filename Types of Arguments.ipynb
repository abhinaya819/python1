```python
#Default arguments: The Parameters defined with some default values , and 
#it assigns the argument value if you pass the values for default parameter otherwise,
# it will assign default values
def sum (a,b,c=0,d=0):
    res = a+b+c+d
    return res
print(sum(10,30))
print(sum(10,30,40))
print(sum(10,30,40,50))
```

```python
def mul (a,b,c=0,d=0):
    res = a*b*c*d
    return res
print(mul(10,30))
print(mul(10,30,40))
print(mul(10,30,40,50))
```
