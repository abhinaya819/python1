```python
#Find the weather the number is prime number or not usins function
def prime_check(n: int) -> bool:
    if n <= 1:
        return False
    if n == 2:
        return True
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

print(prime_check(7)) 

```

```python
#Find the prime numbers 1 to 100
def prime_check(n: int) -> bool:
    if n<2:
        return False
    for i in range(2, n//2+1):
        if n % i == 0:
            return False
    return True
for j in range(1,100):
    if prime_check(j):
        print(j,end=" ")

```
