```python
n=int(input())
res=[]
for i in range(1,n+1):
    res.append(i)
print(res)
```

```python
#List Comprehension: Write the logic inside a list
#Syntax:[exp for iterator in iterable if conditional]
#return 1-n numbers as list using list comprehension
n=int(input())
res=[x for x in range(1,n+1)]
print(res)
```

```python
#Find the first n even numbers
n=int(input())
evens = [x for x in range(0,2*n,1) if x % 2 == 0]
print(evens)
```

```python
#Find List even or odd,for list of integers
arr = [12, 7, 9, 24, 35]
even_odd = [(x, "even" if x % 2 == 0 else "odd") for x in arr]
print(even_odd)
```

```python
# If the number is even add one or number is odd add 0.5
arr = [10, 7, 4, 9, 12, 15]
res = [x + 1 if x % 2 == 0 else x + 0.5 for x in arr]
print(res)
```

```python
# Find one dice combinations
print([x for x in range(1,6)])
```

```python
#Find two dice combinations
print([(i,j) for i in range(1,4) for j in range (1,4)])
```

```python
#Find the two dice combination if any dice face is 4
print([(i,j) for i in range(1,6) for j in range(1,6) if i==4 or j==4])
```
