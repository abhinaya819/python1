```python
# Dictionary ordered collection
details = {'name':'niki', 'age':20, 'course':'python'}
print(details)
```

```python
#dictionary can't allow list as key
#If dictionary contains multiple items with key, In that case it updates first item key value with last defined item value
details = {'name':'niki', 'age':20, 'course':'python', 'marks':[85,90,95],'age':50}
print(details)
```

```python
#assces values  by using keys
details = {'name':'niki', 'age':20, 'course':'python', 'marks':[85,90,95],'age':50}
print(details['course'])
print(details['age'])

# Updating the dictionary value
details['age']+=1
details['course']+'java' # It updates the value for 'course' key
print(details['age'])
print(details)

#item assignment
details['lcation']='india' # It adds new item to dictionary
print(details)
```

```python
### Dictionary Methods
- Get value
 -get(key,default_val) : It returns the value if key exists in dictionary otherwise 'default_val'
- Update values and add items to Dictionary
 - Update(new dictionary): Dictionary Updates with New_Dictionary Items
-Remove items from Dictionary
 -pop(key): It returns value and remove item if key is exists in dictionary otherwise 'KeyError'
 -popitem(): It returns tuple of key value pair and removes the last item from dictionary
-Get keys, value and Items from dictionary
    - Keys(): It returns list of dictionary keys
    - Values(): It returns list of dictionary values
    -items(): It returns list of tuple key-value pairs
```

```python
details = {'name':'niki', 'age':20, 'course':'python', 'marks':[85,90,95],'age':50}
#print(details['location'])
print(details.get('name'))
print(details.get('location'))
print(details.get('age',18))
print(details.get('location',"Location key not exists"))
```

```python
details = {'name':'niki', 'age':20, 'course':'python'}
d = {'marks':[95,85,90], 'age':50}
details.update(d)
print(details)
```

```python
details = {'name':'niki', 'age':20, 'course':'python', 'marks':[85,90,95],'age':50}
print(details.pop('age'))
print(details)
print(details.popitem())
print(details)
#print(details.pop('percentage'))
print(details.keys())
print(details.values())
print(details.items())
print(details)
```

```python
details = {'name':'niki', 'age':20, 'course':'python', 'marks':[85,90,95],'age':50}
print(details.keys())
print(details.values())
print(details.items())
print(details)
```
