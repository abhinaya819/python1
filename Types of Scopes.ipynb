```python
#Scope: In Python scope is region of code , This indicates accessibility of variables or names
        # In Python Accessibility of variables depends on LEGB Role
        # L-Local Scope
        # E-Enclosing Scope
        # G- Global Scope
        # B-Built in Scope
# 1.Local Scope: Define a Variable inside a function those variables called as Local Scope
# We can access the Local Scope variable within the Scope , We can't access in global Scope.

# 2.Enclosing Scope: This Scope can happend through nested functions,
                    # here a variable defind in outer , We can access inside the inner function,
                    # Those variables Scope called as Enclosing Scope.
# 3.Global Scope: The variables are defind outside the all functions,
                # those variables Scope we called as Global Scope.
# 4.Built-in Scope: It checks the names in python built-in Scope.
                
```

```python
#Local Scope
def fun(x):
    x=x+1  # x is Local Scope
    return x
a=10
res = fun(a)
print(res)
```

```python
#Enclosing Scope
def outer(x):
    y=20   #Enclosing Scope
    def inner():
        a=10   #Local Scope
        z=y+a
        print(f"This is a Local Scope:{z}")
    inner()
    y=y+x
    print(f"This is a Enclosing Scope:{y}")
n=5
outer(n)
```

```python
#Enclosing Scope
def outer(x):
    x=10   #Enclosing Scope
    def inner():
        x=20   #Local Scope
        z=x+x
        print(f"This is a Local Scope:{z}")
    inner()
    y=x+x
    print(f"This is a Enclosing Scope:{y}")
n=5
outer(n)
```

```python
#Global Scope
x=10
def outer():
    y=20
    def inner():
        z=30
        res =z+y+x
        print(f"This is Local Scope:{res}")
    res=y+x
    inner()
    print(f"This is Enclosing Scope:{res}")
outer()
```

```python
#Global Scope
x=10 #Global Scope 
def outer():
    y=20  # Enclosing Scope
    def inner():
        z=30  #Local scope
        res =z+y+x
        print(f"This is Local Scope:{res}")
    y=y+x
    inner()
    print(f"This is Enclosing Scope:{res}")
outer()
```

```python
#Trying to update the global scope variable inside the local scope
#Global Scope
x=10 
def outer():
    x=x+1
    print(f"This is Local Scope val:",x)
outer()
print(f"This is Global Scope val:",x)
```

```python
#update the global scope variable inside the local scope by using 'global' keyword
#Global Scope
x=10 
def outer():
    global x
    x=x+1
    print(f"This is Local Scope val:",x)
outer()
print(f"This is Global Scope val:",x)
```

```python
#Trying to update Enclosing Scope variable in Local Scope
x=10 #Global Scope 
def outer():
    y=20  # Enclosing Scope
    def inner():
        z=30  #Local scope
        y=z+1
        print(f"This is Local Scope:{y}")
    y=y+x
    inner()
    print(f"This is Enclosing Scope:{y}")
outer()
```

```python
#Trying to update Enclosing Scope variable in Local Scope using 'nomlocal' keyword
x=10 #Global Scope 
def outer():
    y=20  # Enclosing Scope
    def inner():
        nonlocal y
        z=30  #Local scope
        y=y+1
        print(f"This is Local Scope:{y}")
    y=y+x
    inner()
    y=y+x
    print(f"This is Enclosing Scope:{y}")
outer()
```
