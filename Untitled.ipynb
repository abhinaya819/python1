```python
### String Method
- Case Conversion Methods
    - upper() : It converts entire string into uppercase
    -lower() : It converts entire string into lowercase
    -title() :It converts every word first character into uppercase and remember
    -Swapcase()
    -capitalize()
```

```python
string = "The ocean is blue and sky is Blue"
lower = string.lower()
print(lower)
print(string.upper())
print(string.title())
print(string.swapcase())
print(string.capitalize())
```

```python
#white space remove methods
# strip()
# lstrip()
# rstrip()
string = "      codegnan    "
print(string)
print(string.strip())
print(string.lstrip())
print(string.rstrip())
```

```python
#white space remove methods
# strip()
# lstrip()
# rstrip()
string = "@@@@@@@code  @  gnan@@@@@@@"
print(string)
print(string.strip("@"))
print(string.lstrip("@"))
print(string.rstrip())
```

```python
# split():
string = "The ocean is blue and sky is Blue 1234@"
print(string.split())
print(string.split('is'))
print(string.split('z'))
print(string.split('blue'))
print(string.split('@'))
print(string.split('1'))
```

```python
# Indexing finding methods
# index(sub_string)
# find(sub_string)
string = "The ocean is blue and sky is Blue"
print(string.index("o"))
print(string.index('is'))
print(string.index('is',14)) #it starts checking from 14th index
print(string.index('z'))
```

```python
# Indexing finding methods
# index(sub_string)
# find(sub_string)
string = "The ocean is blue and sky is Blue"
print(string.find("o"))
print(string.find('is'))
print(string.find('is',14)) #it starts checking from 14th index
print(string.find('z'))
```

```python
#replace(old, new): It replae the  old_string with new_string in string
string = "The ocean is blue and sky is Blue"
print(string.replace('blue','red'))
print(string.replace('blue','red', 1)) # It replaces only 1 occurence
```

```python
#"sep".join(iterable):it joins the all iterable elements with seperator
lst = ['The', 'ocean', 'is', 'blue', 'and', 'sky', 'is', 'Blue']
print("".join(lst))
print(" ".join(lst))
print("_".join(lst))
```

```python
#### string checing methods
''''- islower():
- isupper():
- istitle():
- isalpha():
- isalnum():
- isdigit():
- startswith():
- endswith(): '''

print("hello".islower())
print("Hello".islower())
print("HELLO".isupper())
print('123'.isdigit())
print('abcABC'.isalpha())
print('1234'.isalnum())
print('1342abc'.isalnum())
print('Hello World'.startswith('H'))
print('Hello World'.endswith('H'))


```
