```python
#defines set
#set is unordered collection
s = {10,2,4,5,3,8, 'code', 'Hello', 'Niki',3.25}
print(s)
print(hash(10))
print(hash(2))
print(hash(4))
print(hash(3.25))
```

```python
# set return only unique elements
s = {1,1,2,2,3,3,4,5,6,7,8}
print(s)
print(len(s))
```

```python
#set does not allow unhashable values as elements
s = {1,1,2,2,3,3,[4,5,6,7],8}
print(s)
print(len(s))
```

```python
# Set Methods
- Adding Elements To set
    -add(ele): It adds elements to set if elements not exists in set , otherwise nothing.
    -update(iterable): It adds all elements to set
- Remove Elements from set
    -remove(ele): It removes the elmennts from if elements exists in set otherwise 'keyerror'.
    -discard(ele): It returns the elements from set if elements exists in set otherwise noting.
    -Pop(): It remove and returns the elements from set
- Set Operations
    -unique(another_sets): It returns the all common elements from all sets.
    -intersection(another_sets): It returns the common elements from the all sets.
    -difference(anoter_sets): It returns the set elements those are not lies in another_sets.
    -symmetric_difference(another_sets): It returns the non-common elements from all sets.
```

```python
s={1,2,3,4,5,6,-2,-3,-4}
s.add(5)
print(s)
s.update([10,40,50])
print(s)
s.remove(3)
print(s)
#s.remove(3) #keyerror
s.discard(3)
print(s)
```

```python
s1={1,2,3,4,5,6,1}
s2={1,6,7,8,3,2}
s3={9,8,7}
print(s1 | s2) #union
print(s1 | s2 | s3)
print(s1 & s2) #intersection
print(s1 & s2 & s3)
print(s1 - s2) #difference
print(s1 - s2 - s3)
print(s1 ^ s2) #symmetric
print(s1 ^ s2 ^ s3)


```
